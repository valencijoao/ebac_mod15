Como a coluna Genero possui uma porcentagem muito baixa de dados nulos (0,48%), optei por excluir as linhas com valores ausentes nessa coluna. Essa remoção terá impacto mínimo na base de dados e evita preencher os valores com uma informação que não pode ser inferida com segurança.


```python
df.dropna(subset=['Genero'], inplace=True)
```

C) Nessa etapa substitua pela média, mediana ou moda as linhas das colunas com valores faltantes que você analisou e entendeu que esse era o melhor método. Não se esqueça de justificar sua escolha.

**Dica variáveis numéricas:** Não se esqueça de verificar a distrbuição dos dados, média e mediana.

**Dica variáveis categóricas:** Caso precise substituir uma variável categórica utilize um gráfico de barras para verificar a que mais aparece e utilizar ela na substituição.

Caso opte por substituir você pode utilizar a função fillna:
df['SUACOLUNA'].fillna("VALOR QUE SUBSTITUIRA O NULO", inplace=True)
Ela funcionará como a replace.

```python
df["Pagamento_Mensal"] = df["Pagamento_Mensal"].fillna( df["Total_Pago"] / df["Tempo_como_Cliente"].replace(0, pd.NA))
```

A coluna Pagamento_Mensal possui valores nulos e foi preenchida utilizando a relação entre Total_Pago e Tempo_como_Cliente, pois essa fórmula permite estimar o valor mensal pago por cada cliente de forma coerente com a estrutura dos dados. Essa abordagem mantém a consistência da base e evita a perda de registros.

```python
df['PhoneService'] = df['PhoneService'].fillna('Sem informacao')
```

A coluna Servico_Telefonico possui uma alta quantidade de valores nulos. Em vez de excluir a coluna, optei por substituir os valores ausentes por Sem informacao, preservando os registros e evitando perda de dados.
