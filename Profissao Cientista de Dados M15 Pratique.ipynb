Identificação de outliers:

```python
colunas_numericas = [
    "Pagamento_Mensal",
    "Tempo_como_Cliente",
    "Total_Pago"
                     ]

for coluna in colunas_numericas:
    fig = px.box(
        df,
        x=coluna,
        points='outliers',

        title=f'Boxplot interativo - {coluna}',
        labels={coluna: coluna},

    )

    fig.update_xaxes(
        tickprefix="R$ ",
        tickformat=",.2f"
    )

    fig.update_layout(
        separators=",."
    )

    fig.show()
```

```python
colunas_valores = [
    "Pagamento_Mensal",
    "Tempo_como_Cliente",
    "Total_Pago"
]

for coluna in colunas_valores:
    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = df[
        (df[coluna] < limite_inferior) |
        (df[coluna] > limite_superior)
    ] 


    print(f"\nColuna: {coluna}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Limite superior: {limite_superior}")
    print(f"Limite inferior: {limite_inferior}")
    print(f"Outliers: {len(outliers)}")
```

```python
colunas_boolean = [
    "Servico_Seguranca",
    "Suporte_Tecnico",
    "StreamingTV",
    "Casado",
    "Dependentes",
    "Churn"
]

for coluna in colunas_boolean:
    contagem_valores = (
        df[coluna]
        .value_counts()
        .reindex([0,1],fill_value=0)
        .rename_axis('Valor')
        .reset_index(name='Quantidade')
    )

    contagem_valores['Categoria'] = contagem_valores['Valor'].map({
        0:'Não',
        1:'Sim'
    })

    fig = px.bar(
        contagem_valores,
        x='Categoria',
        y='Quantidade',
        text='Quantidade',
        color='Categoria',
        title=f"Distribuição da variável {coluna}",
        labels={
            'Categoria': coluna,
            'Quantidade': 'Quantidade'
        },
        category_orders={
            'Categoria':['Não','Sim']
        }
    )    

    fig.update_traces(textposition='outside')
    fig.update_layout(showlegend=False, autosize=False, width=1200, height=800)


    pd.set_option('display.width', 1000)
    print(f"\n{coluna}")
    print(contagem_valores.to_string(index=False))
    fig.show()
```

*Os gráficos referentes às variáveis booleanas no mostram que só são balanceados na coluna 'Casado', porém todos os registros apresentam um número considerável em ambas as categorias de resposta.*

#### - Questionamento 3: Qual o perfil do cliente que possui dependentes e do que não possui?

```python
def tracar_perfil(coluna):
    """
    Traça o perfil de cliente em relação ao consumo atrelado à família.
    """
    if coluna["Casado"] and coluna["Dependentes"] == 1:
        return "Casado com dependentes"
    elif coluna["Casado"] == 1 and coluna["Dependentes"] == 0:
        return "Casado sem dependentes"
    elif coluna["Casado"] == 0 and coluna["Dependentes"] == 1:
        return "Solteiro com dependentes"
    else:
        return "Solteiro sem dependentes"

df["Perfil_Familia"] = df.apply(tracar_perfil, axis=1).sort_values(ascending=True)

print(df["Perfil_Familia"].value_counts())
```

```python
df[colunas_produtos] = df_contratados

perfil_produtos = (
    df.groupby("Perfil_Familia")[colunas_produtos].mean() * 100
).round(2)

print(perfil_produtos)
```

```python
df_grafico = perfil_produtos.reset_index().melt(
    id_vars="Perfil_Familia", var_name="Produto", value_name="Taxa_Adesao_Pct"
).sort_values(by="Taxa_Adesao_Pct", ascending=True)

fig = px.bar(
    df_grafico,
    x="Perfil_Familia",
    y="Taxa_Adesao_Pct",
    color="Produto",
    barmode="group",
    text="Taxa_Adesao_Pct",
    title="<b>Taxa de Adesão aos Produtos por Perfil Familiar (%)</b>",
    labels={
        "Perfil_Familia": "Perfil Familia",
        "Taxa_Adesao_Pct": "Adesão (%)",
    },
    color_discrete_sequence=px.colors.qualitative.Set2,
)

fig.update_traces(texttemplate="%{text}%", textposition="outside")
fig.update_layout(
    height=500, width=850, yaxis_range=[0, 100], title_x=0.5, legend_title="Produto"
)

fig.show()
```

Percebemos que clientes <b>casados</b> costumam a contratar mais o serviço de entretenimento em questão. Com os clientes casados e sem dependentes com um número ligeiramente maior, porém, em relação aos serviços de manutenção estes clientes são os que mais possuem contratos, o que pode indicar que clientes com mais usuários costumam a optar por um pacote mais completo de produtos.

### Questionamento 4: Qual o modelo de contrato mais sustentável à longo prazo?

```python
resumo_medidas = (
    df.groupby("Churn")[colunas_numericas].mean().round(2).reset_index()
)

print(f"========= Média das variáveis numéricas por Churn =========")
print(resumo_medidas)
```

```python
for coluna in colunas_numericas:
    fig = px.box(
        df,
        x=coluna,
        y="Churn",
        color="Churn",
        title=f"<b>Distribuição de {coluna} por Churn</b>",
        labels={"Churn": "Churn", coluna: coluna},
        color_discrete_sequence=["#636EFA", "#EF553B"],
        orientation="h"
    )
    fig.update_layout(width=1000, height=700, title_x=0.5, showlegend=False)
    fig.show()
```

Os clientes costumam a cancelar o serviço nos primeiros meses, o que acusa um problema de retenção de clientes.

```python
churn_tipo_contrato = (
    df.groupby("Tipo_Contrato")["Churn"]
    .agg(Total_Clientes="count", Churn_Total="sum")
    .reset_index()
)

churn_tipo_contrato["Taxa_Churn_Pct"] = (
    churn_tipo_contrato["Churn_Total"] / churn_tipo_contrato["Total_Clientes"] * 100
).round(2)

print("========= Taxa de Churn por Tipo de Contrato =========")
print(churn_tipo_contrato)
```

```python
fig = px.bar(
    churn_tipo_contrato.sort_values(by="Taxa_Churn_Pct",ascending=True),
    x="Tipo_Contrato",
    y="Taxa_Churn_Pct",
    text="Taxa_Churn_Pct",
    color="Tipo_Contrato",
    title="<b>Taxa de Churn (%) por Tipo de Contrato</b>",
    labels={
        "Tipo_Contrato": "Tipo de Contrato",
        "Taxa_Churn_Pct": "Taxa de Churn (%)",
    },
    color_discrete_sequence=px.colors.qualitative.Safe,
)

fig.update_traces(texttemplate="%{text}%", textposition="outside")
fig.update_layout(height=450, width=700, title_x=0.5, yaxis_range=[0, 100])
fig.show()
```

### Questionamento 5: Quais variáveis possuem a maior correlação com churn?

```python

vertentes = ["Tempo_como_Cliente", "Total_Pago"]

correlacao = (
    df[vertentes + ["Churn"]]
    .corr()["Churn"]
    .drop("Churn")
    .abs()
    .reset_index()
)
correlacao.columns = ["Vertente", "Forca_Relacao"]
correlacao["Vertente"] = correlacao["Vertente"].map(
    {
        "Tempo_como_Cliente": "Tempo como Cliente",
        "Total_Pago": "Valor Total Pago",
    }
)
correlacao = correlacao.sort_values(by="Forca_Relacao", ascending=True)

fig = px.bar(
    correlacao,
    x="Forca_Relacao",
    y="Vertente",
    orientation="h",
    text=correlacao["Forca_Relacao"].apply(lambda x: f"{x:.2f}"),
    color="Forca_Relacao",
    color_continuous_scale="Purples",
    title="<b>Comparativo da Relação das Vertentes com o Churn</b>",
    labels={
        "Forca_Relacao": "Grau de Relação / Associação",
        "Vertente": "Vertente Analisada",
    },
)

fig.update_traces(textposition="outside")
fig.update_layout(
    height=400,
    width=750,
    title_x=0.5,
    xaxis_range=[0, 1.0],
    coloraxis_showscale=False,
)

fig.show()
```
