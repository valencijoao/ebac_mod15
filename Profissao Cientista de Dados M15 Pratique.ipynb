```python
#Seu código aqui
df.describe()
# A) Podemos observar que grande parte dos dados na coluna 'Tempo_como_Cliente' está concentrada no Q3, assim como há uma grande diferença entre o mínimo e o máximo. Com essas observações, podemos esperar a presença de outliers acima da média. O mesmo não se aplica ao 'Total_Pago', onde grande partes do dados se concentram no Q2, porém este valor não condiz com a média, o que nos leva a concluir a presença de outliers. A coluna 'Pagamento_Mensal' possui o valor de desvio pedrão baixo em relação a média, isso nos mostra que os dados estão em sua maioria concentrados ao redor da média, e as medidas de Q1 e Q3 demonstram este comportamento, a semelhança entre média mediana também contribui com o pensamento da presença de poucos ou nenhum outlier.
```

```python
mediana_pagamento = df['Pagamento_Mensal'].median()
mediana_tempo = df['Tempo_como_Cliente'].median()
mediana_total = df['Total_Pago'].median()

print(f"Calculos de Mediana:")
print(f"  Pagamento Mensal: {mediana_pagamento}")
print(f"  Tempo Como Cliente: {mediana_tempo}")
print(f"  Total pago: {mediana_total}")
```

Identificação de outliers:

```python
colunas_numericas = [
    "Pagamento_Mensal",
    "Tempo_como_Cliente",
    "Total_Pago"
                     ]

for coluna in colunas_numericas:
    fig = px.box(
        df,
        x=coluna,
        points='outliers',

        title=f'Boxplot interativo - {coluna}',
        labels={coluna: coluna},

    )

    fig.update_xaxes(
        tickprefix="R$ ",
        tickformat=",.2f"
    )

    fig.update_layout(
        separators=",."
    )

    fig.show()
```

```python
colunas_valores = [
    "Pagamento_Mensal",
    "Tempo_como_Cliente",
    "Total_Pago"
]

for coluna in colunas_valores:
    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = df[
        (df[coluna] < limite_inferior) |
        (df[coluna] > limite_superior)
    ] 


    print(f"\nColuna: {coluna}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Limite superior: {limite_superior}")
    print(f"Limite inferior: {limite_inferior}")
    print(f"Outliers: {len(outliers)}")
```

```python
colunas_boolean = [
    "Servico_Seguranca",
    "Suporte_Tecnico",
    "StreamingTV",
    "Casado",
    "Dependentes",
    "Churn"
]

for coluna in colunas_boolean:
    contagem_valores = (
        df[coluna]
        .value_counts()
        .reindex([0,1],fill_value=0)
        .rename_axis('Valor')
        .reset_index(name='Quantidade')
    )

    contagem_valores['Categoria'] = contagem_valores['Valor'].map({
        0:'Não',
        1:'Sim'
    })

    fig = px.bar(
        contagem_valores,
        x='Categoria',
        y='Quantidade',
        text='Quantidade',
        color='Categoria',
        title=f"Distribuição da variável {coluna}",
        labels={
            'Categoria': coluna,
            'Quantidade': 'Quantidade'
        },
        category_orders={
            'Categoria':['Não','Sim']
        }
    )    

    fig.update_traces(textposition='outside')
    fig.update_layout(showlegend=False, autosize=False, width=1200, height=800)


    pd.set_option('display.width', 1000)
    print(f"\n{coluna}")
    print(contagem_valores.to_string(index=False))
    fig.show()
```

*Os gráficos referentes às variáveis booleanas no mostram que só são balanceados na coluna 'Casado', porém todos os registros apresentam um número considerável em ambas as categorias de resposta.*
